# Tuple Advanced Practice

This notebook covers important tuple concepts you can practice after the basics.

```python
# 1) Indexing and slicing
numbers = (10, 20, 30, 40, 50, 60)
print(numbers[1])
print(numbers[1:4])
print(numbers[-3:])
```

```python
# 2) Tuple unpacking
point = (3, 5)
x, y = point
print(x, y)
```

```python
# 3) Count and index
data = (1, 2, 2, 3, 2, 4)
print(data.count(2))
print(data.index(3))
```

```python
# 4) Converting tuple to list and back
t = (1, 2, 3)
lst = list(t)
lst.append(4)
t2 = tuple(lst)
print(lst)
print(t2)
```

```python
# 5) Iterating through a tuple
colors = ('red', 'green', 'blue')
for color in colors:
    print(color)
```

```python
# 6) Nested tuples
nested = ((1, 2), (3, 4), (5, 6))
print(nested[1][0])
for item in nested:
    print(item)
```

```python
# 7) Tuple comprehension (simple example)
squares = tuple(x * x for x in range(1, 6))
print(squares)
```
