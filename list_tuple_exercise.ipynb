# List and Tuple Practice Questions

Try to solve these simple logical exercises on lists and tuples.

## 1. Find the first and last item
Create a list called `fruits` with 4 items. Print the first and last item.

## 2. Add an item to a list
Create a list `numbers = [1, 2, 3]` and add `4` to the end.

## 3. Remove an item from a list
Create a list `colors = ['red', 'green', 'blue']` and remove `'green'`.

## 4. Count how many times a value appears
In the list `[1, 2, 2, 3, 2]`, count how many times `2` appears.

## 5. Reverse a list
Create a list `[10, 20, 30, 40]` and print it in reverse order.

## 6. Create a tuple and access one item
Create a tuple `days = ('Mon', 'Tue', 'Wed')` and print the second item.

## 7. Convert a list to a tuple
Convert the list `[1, 2, 3]` into a tuple.

## 8. Find the length
Create a tuple `point = (4, 5, 6)` and print its length.

## 9. Bonus logical thinking
Given `data = [5, 8, 3, 8, 1]`, print the number of times `8` appears and the sum of all values.
