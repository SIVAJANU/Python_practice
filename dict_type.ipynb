```python
# Dictionary example: store key-value pairs
person = {
    "name": "Alice",
    "age": 30,
    "city": "Seattle"
}

# Access values by key in dictionary and in case of list we access the values by index list_1[0]
print(person["name"])  # Alice 

# Add or update items or keys or update values of the keys
person["email"] = "alice@example.com"
person["age"] = 31

# Iterate over keys and values
for key, value in person.items():
    print(key, value)

# Use case: lookup data by a unique identifier
phone_book = {
    "Bob": "206-555-0100",
    "Carol": "206-555-0111"
}
print(phone_book.get("Carol"))  # 206-555-0111


# Set example: store unique elements
fruits = {"apple", "banana", "orange"}

# Add and remove items
fruits.add("kiwi")
fruits.discard("banana")

# Use set operations
tropical = {"banana", "kiwi", "mango"}
print('intersection use case of set data type',fruits & tropical)  # intersection
print('union use case of set data type',fruits | tropical)  # union

# Use case: remove duplicates and test membership efficiently
numbers = [1, 2, 2, 3, 3, 3]
unique_numbers = set(numbers)
print(unique_numbers)  # {1, 2, 3}
print(2 in unique_numbers)  # True
```

```python
# Dictionary example: store key-value pairs
person = {
    "name": "Alice",
    "age": 30,
    "city": "Seattle"
}

# Access values by key in dictionary
print(person["name"])  # Alice

# Add or update items
person["email"] = "alice@example.com"
person["age"] = 31

# Iterate over keys and values
for key, value in person.items():
    print(key, value)

# Use case: lookup data by a unique identifier
phone_book = {
    "Bob": "206-555-0100",
    "Carol": "206-555-0111"
}

# Using .get() to safely access values
print(phone_book.get("Carol"))              # 206-555-0111
print(phone_book.get("David"))              # None
print(phone_book.get("David", "Not found"))  # Not found

# Compare with direct indexing
# print(phone_book["David"])  # This would raise KeyError
```

```python
# If you want to find the duplicate values from a list and store them in a new variable, use this:
from collections import Counter

numbers = [1, 2, 2, 3, 3, 3, 4]

count = Counter(numbers)
duplicates = {x for x, c in count.items() if c > 1}

print(duplicates)   # {2, 3}
```
