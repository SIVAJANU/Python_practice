```python
def greet(name):
    return f"Hello, {name}!"

print(greet("Asha"))
print(greet("Ravi"))
```

## 2. Parameters and return values

```python
def add(a, b):
    return a + b

result = add(10, 20)
print("Sum:", result)
```

## 3. Default and keyword arguments

```python
def greet_user(name="Guest", age=18):
    print(f"Welcome, {name}! You are {age} years old.")

greet_user()
greet_user("Nisha")
greet_user(age=25)
```

```python
def add(*args):
    return sum(args)

result = add(10,20,30,100)
print("Sum:", result)
```

## 4. Variable-length arguments: *args and **kwargs

```python
def add_all(*args):
    total = 0
    for num in args:
        total += num
    return total

print(add_all(1, 2, 3))
print(add_all(10, 20, 30, 40))

def show_info(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_info(name="Maya", age=22, city="Hyderabad", adress='miyapur')
```

## 5. Lambda functions

```python
# syntax for lambda function
# lambda arguments: expression
square = lambda x: x ** x #anonymous function 5*5*5*5*5
print(square(5))

add_two = lambda a, b: a + b
print(add_two(10, 17))
```

## 6. Scope: Local, Global, and nonlocal

```python
x = "global"
 # practice for local, global and nonlocal variables
def test_local():
    x = "local"
    print("Inside local function:", x)

def test_global():
    print("Inside global function:", x)

def outer():
    count = 10

    def inner():
        count += 1
        print("Inside inner function:", count)

    inner()
    print("Inside outer function:", count)

test_local()
test_global()
outer()
```

## 7. Recursion

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

## 8. Practice exercises

```python
# Exercise 1: Write a function to check if a number is even
def is_even(number):
    return number % 2 == 0

print(is_even(4))
print(is_even(7))

# Exercise 2: Write a function that accepts any number of values and returns their sum
def total(*numbers):
    result = 0
    for num in numbers:
        result += num
    return result

print(total(2, 3, 4))
print(total(1, 2, 3, 4, 5))
```
