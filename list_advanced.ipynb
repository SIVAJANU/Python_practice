```python
# 3) Sorting and reversing
nums = [5, 1, 9, 3, 7]
nums.sort()
print('Sorted:', nums)
nums.reverse()
print('Reversed:', nums)
print(nums[::-1])
```

```python
# 4) Copying a list
original = [1, 2, 3]
copy1 = original.copy()
copy2 = list(original)
copy1.append(4)
print('Original:', original)
print('Copy1:', copy1)
print('Copy2:', copy2)
```

```python
# 5) Membership and counting
letters = ['a', 'b', 'c', 'a', 'd']
print('a in letters:', 'a' in letters)
print('Count of a:', letters.count('a'))
```

```python
list_nums=['1',1,1,1,11,11,22,33,22,2,2,2]
list_nums.count(1)
```

```python
# 6) List comprehension
squares = [x * x for x in range(1, 8)]
print(squares)
even_numbers = [x for x in range(1, 11) if x % 2 == 0]
print(even_numbers)
```

```python
# 7) Nested lists
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print(matrix[1][2]) # [4, 5, 6] --->6
for row in matrix:
    print(row)
```

```python
# 8) Unpacking and enumerate
colors = ['red', 'green', 'blue'] # index and value
# index of red color is 0
# index of green color is 1
# index of blue color is 2
# a, b, c = colors
# print(a, b, c)
for index, color in enumerate(colors):
    print(index, color)
```

```python
for i in colors:
    print(i)
```

```python
L=len(colors)
print(L)
for i in range(L):
    print(colors[i])
```
