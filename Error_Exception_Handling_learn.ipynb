````python
try:
    with open('sample1.txt','w') as file_text:
        file_text.write('Hello today you are doing good')
    with open('sample1.txt','r') as file_text:
        content=file_text.read()
        print(content)
except Exception as err:
    print("Error that is to be corrected is: ", err)
finally:
    print("This block always runs.")
````

## 7. Raising exceptions with raise

````python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older.")
    return "Eligible"

try:
    print(check_age(90))
except ValueError as e:
    print(e)
````

````python
## Using `raise` in Python

`raise` is used when you want to trigger an exception intentionally.

Common cases:
- When validating input or state and you want to stop execution with a clear error.
    - Example: `if age < 18: raise ValueError("Age must be 18 or older.")`
- When you want to use a custom exception type for a specific error condition.
    - Example: `raise InvalidAgeError("Age must be 18 or older.")`
- When you need to re-raise the current exception after doing some cleanup or logging.
    - Example:
        ```python
        except Exception:
                # do something
                raise
        ```
- When wrapping one exception inside another to preserve context.
    - Example: `raise MyError("failed") from original_exc`
- When a function or module detects an unrecoverable condition and must notify the caller.

In short, use `raise` whenever you need your code to explicitly signal an error instead of letting the interpreter fail later without a clear message.
````

````python
try:
    for i in range(1,100):
        print(i)
        if i >=10:
            raise ValueError('The number is more than 10')
except Exception as err:
    print(err)
````

## 8. Custom exceptions

````python
class InvalidAgeError(Exception):
    pass

def verify_age(age):
    if age < 18:
        raise InvalidAgeError("Age must be 18 or older.")
    return "Valid age"

try:
    print(verify_age(16))
except InvalidAgeError as e:
    print("Custom exception caught:", e)
````
