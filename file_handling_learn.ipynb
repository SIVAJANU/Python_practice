## 5. File modes: r, w, a, rb, wb

```python
# r -> read
# w -> write (overwrites existing content)
# a -> append
# rb -> read binary
# wb -> write binary

with open("sample.txt", "r") as f:
    print("Text mode read:", f.read())

with open("sample.bin", "wb") as f:
    f.write(b"Binary data example")

with open("sample.bin", "rb") as f:
    print("Binary mode read:", f.read())
```

## 6. The with statement

```python
with open("sample.txt", "r") as f:
    lines = f.readlines()

for line in lines:
    print(line.strip())
```

## 7. Why we need exception handling

```python
try:
    with open("missing_file.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("The file does not exist. Please check the filename.")
```

```python
try:
    with open('rrrr.txt','r') as text_file:
        print(text_file.read())
except FileNotFoundError:
    print("error: File not found")

print('hi how are you?')
```

# list of File Exception handling errors
# like FileNotFoundError


```python
# Display common file handling exceptions
exceptions = {
    "FileNotFoundError": "Raised when a file or directory is requested but doesn't exist",
    "PermissionError": "Raised when trying to run an operation without the adequate access rights",
    "IsADirectoryError": "Raised when a file operation is requested on a directory",
    "NotADirectoryError": "Raised when a directory operation is requested on a non-directory file",
    "IOError": "Raised when an input/output operation fails",
    "OSError": "Base class for system-related errors",
    "UnicodeDecodeError": "Raised when decoding a string fails",
    "UnicodeEncodeError": "Raised when encoding a string fails"
}

for exception, description in exceptions.items():
    print(f"{exception}: {description}")
```

```python
try:
    for exception, description in exceptions.items():
        print(f"{exception}: {description}")
except Exception as e:
    print(f"An error occurred: {e}")
finally:
    print("Finished displaying exceptions.")
```

## 8. Exception handling with writing

```python
try:
    with open("sample.txt", "w") as f:
        f.write("Writing safely with exception handling.")
except PermissionError:
    print("You do not have permission to write this file.")
except Exception as e:
    print("An unexpected error occurred:", e)
```
