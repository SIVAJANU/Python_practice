# Answers for List and Tuple Practice Questions

These are the solution codes for the practice notebook.

```python
# 1. Find the first and last item
fruits = ['apple', 'banana', 'mango', 'orange']
print(fruits[0])
print(fruits[-1])
```

```python
# 2. Add an item to a list
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)
```

```python
# 3. Remove an item from a list
colors = ['red', 'green', 'blue']
colors.remove('green')
print(colors)
```

```python
# 4. Count how many times a value appears
values = [1, 2, 2, 3, 2]
print(values.count(2))
```

```python
# 5. Reverse a list
nums = [10, 20, 30, 40]
nums.reverse()
print(nums)
```

```python
# 6. Create a tuple and access one item
days = ('Mon', 'Tue', 'Wed')
print(days[1])
```

```python
# 7. Convert a list to a tuple
numbers = [1, 2, 3]
tup = tuple(numbers)
print(tup)
```

```python
# 8. Find the length
point = (4, 5, 6)
print(len(point))
```

```python
# 9. Bonus logical thinking
data = [5, 8, 3, 8, 1]
print(data.count(8))
print(sum(data))
```
