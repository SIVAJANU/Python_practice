````python
### Boolean use cases with Python — examples

1. Comparisons (==, !=, <, >, <=, >=)
```python
age = 20
is_adult = age >= 18
print(is_adult)  # True
```

2. Logical operators (and, or, not)
```python
x = 5
in_range = (x > 0) and (x < 10)
print(in_range)  # True
```

3. Truthiness of objects (empty vs non-empty)
```python
items = []
if items:
    print("has items")
else:
    print("empty")  # prints "empty"
```

4. Membership (in / not in)
```python
s = "hello"
print('h' in s, 'z' not in s)  # True True
```

5. Identity vs equality (is vs ==)
```python
a = [1,2]
b = a
c = [1,2]
print(a == c, a is c, a is b)  # True False True
```

6. Short-circuit evaluation
```python
def side_effect():
    print("called")
    return True

# right side not evaluated because left is False
res = False and side_effect()  
print("res:", res)  # "res: False" (no "called")
```

7. any() / all() for multiple conditions
```python
vals = [0, 1, 2]
print(any(vals), all(v >= 0 for v in vals))  # True True
```

8. Filtering with boolean expressions (list comprehensions)
```python
nums = range(10)
evens = [n for n in nums if n % 2 == 0]
print(evens)  # [0,2,4,6,8]
```

9. Assertions and sanity checks
```python
x = 10
assert x > 0, "x must be positive"
```

10. Converting to boolean with bool()
```python
print(bool(""), bool("text"), bool(0), bool(5))  # False True False True
```
````
