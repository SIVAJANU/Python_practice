# Modules and Packages in Python

This notebook introduces modules, packages, built-in libraries, and external packages with practical examples.

## 1. Why we need modules and packages

```python
# Modules help organize code and reuse logic.
# Packages help group related modules together.

print("A module lets you reuse code without rewriting it.")
```

```python
import pandas as pd
df = pd.DataFrame({"col1": [1, 2], "col2": [3, 4], "col3": [5, 6]})
df
# (2, 3)
```

## 2. Importing modules

```python
import math

print(math.sqrt(16))
print(math.pi)
```

## 3. Built-in module: math

```python
import math

print(math.factorial(5))
print(math.ceil(3.2))
print(math.floor(3.8))
```

## 4. Built-in module: datetime

```python
from datetime import datetime, timedelta

now = datetime.now()
print("Current time:", now)
n=7
param='weeks'
future = now + timedelta(weeks=n)
print(f"After {n} {param}:{future}")
```

## 5. Built-in module: random

```python
import random

print(random.randint(1, 1011))
print(random.choice(["apple", "banana", "cherry",'siva']))
```
