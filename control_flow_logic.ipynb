```python
# while loop # nested condtion
count=0
num=100
while num>10: # ever true logic dont give here in While loop means give a condition that will 
    #fail in any iteration    
    print('The num value is les than 10', num)
    num+=1 # num=num+1
    count+=1
    if num==115:
        print('how many iterations are done',count)
        break
```
