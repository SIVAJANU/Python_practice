# Data Handling & Libraries

This notebook covers:
- Working with JSON
- Working with CSV
- NumPy basics
- Pandas basics
- Matplotlib basics


```python
import json
import csv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# JSON example
json_data = {"name": "Alice", "age": 28, "city": "Delhi"}
with open('sample_data.json', 'w', encoding='utf-8') as f:
    json.dump(json_data, f)

with open('sample_data.json', 'r', encoding='utf-8') as f:
    print(json.load(f))

# CSV example
with open('sample_data.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Alice', 28, 'Delhi'])
    writer.writerow(['Bob', 32, 'Mumbai'])

with open('sample_data.csv', newline='', encoding='utf-8') as f:
    print(list(csv.reader(f)))

# NumPy example
arr = np.array([1, 2, 3, 4, 5])
print('NumPy mean:', arr.mean())

# Pandas example
students = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Marks': [88, 76, 95]
})
print(students)

# Matplotlib example
plt.plot([1, 2, 3, 4], [10, 20, 25, 30])
plt.title('Sample Plot')
plt.show()
```
