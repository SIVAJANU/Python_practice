## 1. Iterators & Generators with yield

A generator is a lazy sequence. It computes values on demand instead of storing them all in memory.

### Custom iterator

A custom class can implement `__iter__` and `__next__` to behave like an iterator.


````python
class MyIterator:
    def __init__(self, data):
        self.data = data
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= len(self.data):
            raise StopIteration
        value = self.data[self.index]
        self.index += 1
        return value

it = MyIterator([10, 20, 30])
for item in it:
    print(item)

# Generator using yield

def square_numbers(n):
    for i in range(1, n + 1):
        yield i * i

print(list(square_numbers(5)))
````

## 2. Decorators

A decorator wraps a function and changes or extends its behavior without modifying the original function body.

```
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished {func.__name__}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

print(add(2, 3))
```


````python
def logger(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished {func.__name__}")
        return result
    return wrapper

@logger
def add(a, b):
    return a + b

add(5, 3)
````

````python
import time
from functools import wraps

def timer(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.4f} seconds")
        return result
    return wrapper

@timer
def expensive_task():
    time.sleep(1)
    return "Task complete"

print(expensive_task())
````

````python
def dec_func(func):
    def wrapper():
        print('Hi you are now entered the decorator function')
        func()        
    return wrapper

@dec_func
def say_hello():
    print("Hello, World!")  

say_hello()  
````

## 3. Context Managers

The `with` statement ensures resources are properly cleaned up after use.

Example: a custom class with `__enter__` and `__exit__`.


````python
class FileManager:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode

    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file

    def __exit__(self, exc_type, exc_value, traceback):
        self.file.close()

with FileManager('example.txt', 'w') as f:
    f.write('Hello from context manager!')

print('File created successfully.')
````

## 4. Regular Expressions

The `re` module supports pattern matching, substitution, and validation.


````python
import re

text = "Contact amit@example.com and support@company.org"
pattern = r"[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}"

matches = re.findall(pattern, text)
print(matches)

print(bool(re.fullmatch(pattern, 'amit@example.com')))
print(re.sub(r'\s+', ' ', 'Python   regex   example'))
````

## 5. Logging

Python's `logging` module helps track application behavior and debug issues.


````python
import logging

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    filename='app.log'
)

logging.info('Application started')
logging.warning('Warning: low disk space')
logging.error('An error occurred')

print('Check app.log for log entries.')
````

## 7. Multithreading & Multiprocessing

Threading is good for I/O-bound work, while multiprocessing is better for CPU-bound tasks.


````python
import threading
import multiprocessing
import time


def worker(name):
    print(f"Thread {name} started")
    time.sleep(1)
    print(f"Thread {name} finished")

threads = [threading.Thread(target=worker, args=(i,)) for i in range(3)]
for t in threads:
    t.start()
for t in threads:
    t.join()


def square(x):
    return x * x

if __name__ == '__main__':
    with multiprocessing.Pool(processes=3) as pool:
        print(pool.map(square, [1, 2, 3, 4, 5]))
````

## 8. Memory Management & Garbage Collection

Python uses reference counting and cyclic garbage collection to manage memory automatically.

The `gc` module can help inspect and trigger collection.


````python
import gc

# Reference counting example
x = [1, 2, 3]
y = x
print('x is y:', x is y)

# Delete x and force collection
x = None
gc.collect()
print('Garbage collector ran successfully.')

# Show gc stats
print('Garbage collector enabled:', gc.isenabled())
````
