```python
def calculate_percentage(score, total):
    # The score must be 0 or greater
    assert score >= 0, "Score cannot be a negative number!"
    
    return (score / total) * 100

# This works fine:
print(calculate_percentage(80, 100))  # Outputs: 80.0

# This will crash the program with your error message:
print(calculate_percentage(-5, 100))
```
