# List and Tuple Bonus Practice

This notebook contains mixed exercises that are useful for teaching and for practicing the difference between lists and tuples.

```python
# 1) Compare mutability
lst = [1, 2, 3]
tup = (1, 2, 3)
lst.append(4)
print('List after append:', lst)
# tup.append(4)  # This will raise an error
print('Tuple stays same:', tup)
```

```python
# 2) Same values, different behavior
a = [1, 2, 3]
b = [1, 2, 3]
c = (1, 2, 3)
d = (1, 2, 3)
print(a == b)
print(c == d)
```

```python
# 3) Convert between list and tuple
numbers = [10, 20, 30]
t = tuple(numbers)
print('Tuple:', t)
l = list(t)
print('List:', l)
```

```python
# 4) Practice indexing on both
sample_list = ['a', 'b', 'c']
sample_tuple = ('x', 'y', 'z')
print(sample_list[1])
print(sample_tuple[-1])
```

```python
# 5) Practice slicing on both
print(sample_list[0:2])
print(sample_tuple[1:])
```

```python
# 6) Iteration and counting
items = ['apple', 'banana', 'apple']
item_tuple = ('apple', 'banana', 'apple')
print(items.count('apple'))
print(item_tuple.count('apple'))
```

```python
# 7) Mixed exercise: build a small record
student = ['John', 21, 'Python']
student_info = ('John', 21, 'Python')
print(student[0], student[1])
print(student_info[0], student_info[1])
```

```python
# 8) Bonus challenge: explain which one is better for fixed data
fixed_data = ('Monday', 'Tuesday', 'Wednesday')
print('This is a good example of tuple usage:', fixed_data)
```
